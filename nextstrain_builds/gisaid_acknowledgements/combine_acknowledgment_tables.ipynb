```python
import pandas as pd
```

```python
viruses = ['h3n2', 'h1n1pdm', 'vic', 'yam']
segments = ['ha', 'na', 'pa', 'pb1', 'pb2', 'mp', 'np', 'ns']

all_tables = []

for v in viruses:
    for s in segments:
        table_filename = f'nextstrain_groups_blab_{v}_{s}_egg_acknowledgements.tsv'
        df = pd.read_csv(table_filename, sep='\t')
        # add columns for virus and segment
        df['virus'] = v
        df['segment'] = s
        all_tables.append(df)
        
# concatenate all DataFrames
combined_tables = pd.concat(all_tables, ignore_index=True)

# save combined table
combined_tables.to_csv('gisaid_acknowledgements.tsv', sep='\t', index=False)
```
